```python
from shapely.geometry import Polygon
import ast

def get_centroid(coord_str):
    try:
        coords = ast.literal_eval(coord_str)
        # En caso de que esté en formato multipolígono (con doble corchete)
        coords_flat = coords[0] if isinstance(coords[0][0], list) else coords
        poly = Polygon(coords_flat)
        centroid = poly.centroid
        return centroid.x, centroid.y
    except Exception as e:
        print(f"Error procesando: {coord_str}\n{e}")
        return None, None
```

```python
df[['centroide_x', 'centroide_y']] = df['place.bounding_box.coordinates'].apply(
    lambda x: pd.Series(get_centroid(x))
)
```
