```python
!pip install --quiet pymupdf

import requests
import fitz  # PyMuPDF

# URL del PDF
url = "https://www.savethechildren.es/sites/default/files/imce/docs/acoso_escolar_y_ciberacoso_informe_vok_-_05.14.pdf"
pdf_filename = "informe.pdf"

# Descargar PDF
response = requests.get(url)
with open(pdf_filename, "wb") as f:
    f.write(response.content)

# Leer el PDF
doc = fitz.open(pdf_filename)
text = ""
for page in doc:
    text += page.get_text()

# Vista previa del contenido
print(text[:1000])
```

```python
import re

# Lista de raíces que nos interesan (puedes agregar más)
raíces = ["acoso", "bully", "hostig", "intimid", "chant", "dox", "groom", "trol", "extorsi", "víctim", "suplant", "difam"]

# Extraer palabras largas del texto
palabras = set(re.findall(r'\b[a-záéíóúñ]{5,}\b', text.lower()))

# Filtrar por aquellas que contienen alguna raíz
lexicon = sorted({w for w in palabras if any(r in w for r in raíces)})

print(f"🔍 Términos extraídos: {len(lexicon)}")
print(lexicon[:20])  # Primeros 20 términos del lexicón
```
